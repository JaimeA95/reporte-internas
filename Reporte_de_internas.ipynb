```python
import pandas as pd

# Ruta del archivo
archivo = "/content/drive/MyDrive/Automatizacion_Reportes/Internas/Junio.xlsx"

# Leer hoja "Data"
df = pd.read_excel(archivo, sheet_name="Data")

# Criterios válidos
criterios = [
    "Instal interna residencial no gravada",
    "Campaña instalacion interna",
    "Instal interna residencial gravada",
    "Instal interna resid reinst no gravada",
    "Instal interna resid reinst gravada",
    "Instal interna pyme no gravada",
    "Instal interna pyme gravada",
    "Instal interna reinst gravada comercial"
]

# Filtrar: criterios válidos + cuenta contrato no nula + estado concluido
df_filtrado = df[
    df["Denomin.posición"].isin(criterios) &
    df["Cuenta contrato"].notna() &
    df["Descr.Status Ttotal"].str.lower().str.contains("concluido")
]

# Nuevo filtro por Plazo > 1
df_filtrado = df_filtrado.copy()
df_filtrado["Plazo"] = pd.to_numeric(df_filtrado["Plazo"], errors="coerce").fillna(1)

# Resolver clase doc.ventas preferida por cuenta contrato
clases = df_filtrado.groupby("Cuenta contrato")["Clase doc.ventas"].unique().reset_index()

def preferida(lista):
    if "ZP06" in lista:
        return "ZP06"
    else:
        return lista[0]

clases["Clase preferida"] = clases["Clase doc.ventas"].apply(preferida)

# Obtener Organización ventas más frecuente por cuenta contrato
org_ventas = df_filtrado.groupby("Cuenta contrato")["Organización ventas"].agg(lambda x: x.mode().iloc[0]).reset_index()

# Unir clase preferida y org. ventas
final = clases.merge(org_ventas, on="Cuenta contrato")
final = final[["Cuenta contrato", "Clase preferida", "Organización ventas"]]

# Renombrar clases doc.ventas
final["Clase doc.ventas"] = final["Clase preferida"].replace({
    "ZCOM": "Mercado Comercial",
    "ZCNE": "Nueva Edificación"
})
final["Clase doc.ventas"] = final["Clase doc.ventas"].apply(
    lambda x: x if x in ["Mercado Comercial", "Nueva Edificación"] else "Saturación"
)

# Agrupar y contar cuentas contrato únicas
resumen = final.groupby(["Organización ventas", "Clase doc.ventas"])["Cuenta contrato"].nunique().reset_index()
resumen.columns = ["Organización ventas", "Clase doc.ventas", "Cuentas contrato únicas"]

# Crear multiíndice para visualización jerárquica
resumen_sin_total = resumen

# Reordenar columnas y aplicar multiíndice
resumen_multi = resumen_sin_total.set_index(["Clase doc.ventas", "Organización ventas"])

# Orden personalizado para Organización ventas
orden_ventas = [15, 24, 18, 294]

# Asegurar tipo entero para ordenar
resumen_multi = resumen_multi.reset_index()
resumen_multi["Organización ventas"] = resumen_multi["Organización ventas"].astype(int)

# Ordenar por Clase doc.ventas y luego por el orden específico
resumen_multi["Organización ventas"] = pd.Categorical(
    resumen_multi["Organización ventas"],
    categories=orden_ventas,
    ordered=True
)
resumen_multi = resumen_multi.sort_values(["Clase doc.ventas", "Organización ventas"])

# Volver a poner MultiIndex
resumen_multi = resumen_multi.set_index(["Clase doc.ventas", "Organización ventas"])

# Crear subtotales por Clase doc.ventas
subtotales = resumen_multi.groupby("Clase doc.ventas")["Cuentas contrato únicas"].sum().reset_index()
subtotales["Clase doc.ventas"] = subtotales["Clase doc.ventas"].apply(lambda x: f"🔸 Total {x}")
subtotales["Organización ventas"] = ""

# Combinar con resumen principal
resumen_multi = pd.concat([resumen_multi.reset_index(), subtotales], ignore_index=True)

# Ordenar primero por Clase doc.ventas con total al final de cada grupo
def orden_clase(x):
    return x.replace("🔸 Total ", "") if "🔸 Total" in x else x

resumen_multi["Clase base"] = resumen_multi["Clase doc.ventas"].apply(orden_clase)
resumen_multi["Es total"] = resumen_multi["Clase doc.ventas"].str.contains("🔸 Total")

resumen_multi = resumen_multi.sort_values(by=["Clase base", "Es total", "Organización ventas"], ascending=[True, True, True])

# Limpiar columnas auxiliares y volver a MultiIndex
resumen_multi = resumen_multi.drop(columns=["Clase base", "Es total"])
resumen_multi = resumen_multi.set_index(["Clase doc.ventas", "Organización ventas"])

# Calcular total general solo desde filas originales (sin subtotales ni total general previos)
resumen_solo_sociedades = resumen_multi[
    ~resumen_multi.index.get_level_values("Clase doc.ventas").str.startswith("🔸 Total")
    & (resumen_multi.index.get_level_values("Clase doc.ventas") != "🔢 Total general")
]

total_general = resumen_solo_sociedades["Cuentas contrato únicas"].sum()

fila_total_general = pd.DataFrame({
    "Clase doc.ventas": ["🔢 Total general"],
    "Organización ventas": [""],
    "Cuentas contrato únicas": [total_general]
})

# Agregar la fila final sin volver a sumar después
resumen_multi = resumen_multi.reset_index()
resumen_multi = pd.concat([resumen_multi, fila_total_general], ignore_index=True)
resumen_multi = resumen_multi.set_index(["Clase doc.ventas", "Organización ventas"])

# Mostrar la tabla ordenada con total general incluido
print(resumen_multi)

# Excluir filas que son subtotales por clase o total general
resumen_solo_sociedades = resumen_multi[
    ~resumen_multi.index.get_level_values("Clase doc.ventas").str.startswith("🔸 Total")
    & (resumen_multi.index.get_level_values("Clase doc.ventas") != "🔢 Total general")
]

total_general = resumen_solo_sociedades["Cuentas contrato únicas"].sum()


fila_total_general = pd.DataFrame({
    "Clase doc.ventas": ["🔢 Total general"],
    "Organización ventas": [""],
    "Cuentas contrato únicas": [total_general]
})

# ➕ Agregar al final del DataFrame
resumen_multi = pd.concat([resumen_multi.reset_index(), fila_total_general], ignore_index=True)

# Volver a poner el índice para mostrar como tabla jerárquica
resumen_multi = resumen_multi.set_index(["Clase doc.ventas", "Organización ventas"])
```
